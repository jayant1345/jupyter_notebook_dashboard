Demonstrates data processing with pandas.

```python
import time
import random
print('Loading data...')
time.sleep(0.5)
```

```python
data = [random.gauss(50, 15) for _ in range(100)]
mean_val = sum(data) / len(data)
print(f'Dataset: {len(data)} rows, mean={mean_val:.2f}')
time.sleep(0.5)
```
